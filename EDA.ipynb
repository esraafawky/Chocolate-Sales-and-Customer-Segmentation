```python
plt.figure(figsize=(8,5))
sns.countplot(data=purchase_df[purchase_df['Incidence'] == 1], x='Brand', color='#8B4513')
plt.title("Number of Purchases (Incidence = 1) per Brand")
plt.xlabel("Brand")
plt.ylabel("Count of Purchases")
plt.show()
```

```python
purchases = purchase_df[purchase_df['Incidence'] == 1]
purchases['Promo_status'] = purchases.apply(lambda row: row[f'Promotion_{int(row.Brand)}'], axis=1)

plt.figure(figsize=(10,6))
sns.countplot(data=purchases, x='Brand', hue='Promo_status', palette=["#D2B48C", "#8B4513"])
plt.title("Purchases per Brand (Promo vs No Promo)")
plt.xlabel("Brand")
plt.ylabel("Number of Purchases")
plt.legend(title="Promotion", labels=["No Promo", "Promo"])
plt.show()
```

```python
inc_counts = purchase_df['Incidence'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(inc_counts, labels=["No Purchase", "Purchase"], autopct="%.1f%%", colors=["#D2B48C", "#8B4513"], startangle=90)
plt.title("Incidence vs No Incidence")
plt.show()
```
